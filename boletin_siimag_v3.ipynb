```python
# ── 5. TEXTO INTRODUCTORIO CON GEMINI ────────────────────────────────────────
resumen = indicadores[['nombre','total','total_comp','porcentaje','tendencia']].to_string(index=False)

instruccion = f"""
Actúa como un analista de datos. Devuelve un texto breve con un análisis de la siguiente
tabla de indicadores académicos:

{resumen}

Menciona los principales cambios entre {PERIODO_COMP_LABEL} y {PERIODO_LABEL}.
No inventes datos. Usa lenguaje formal y sin valoraciones. No lances conclusiones.
El texto debe empezar con: 'En {PERIODO_LABEL}, los principales indicadores académicos
tuvieron el siguiente comportamiento:'
"""

analisis_final = ai.generate_text(instruccion)

def format_miles(texto):
    def reemplazo(match):
        num = match.group()
        if 1900 <= int(num) <= 2099:
            return num
        return f"{int(num):,}"
    return re.sub(r'\d{4,}', reemplazo, texto)

analisis_final   = format_miles(analisis_final)
texto_intro      = analisis_final.replace(". ", ".<br><br>")
print(texto_intro)
```

```python
# ── 6. FUNCIONES AUXILIARES ───────────────────────────────────────────────────

CATEGORIAS = {1: 'Captación', 2: 'Retención', 3: 'Bajas y deserción'}

# Paleta institucional
AZUL       = '#0033A0'
FONDO      = '#f5f7fa'
SEP        = '#e8edf5'
VERDE      = '#388E3C'
VERDE_NUM  = '#1a6b1a'
ROJO       = '#D32F2F'
GRIS       = '#AAA'
GRIS_NUM   = '#555'

BG_VERDE   = '#E8F5E9'
TX_VERDE   = '#2E7D32'
BG_ROJO    = '#FFEBEE'
TX_ROJO    = '#C62828'

def fmt_valor(row):
    v = row['total']
    if row['porcentual'] == 1:
        return f"{v:,.1f}", '%'
    elif v >= 1000:
        return f"{int(v):,}", ''
    return f"{int(v)}", ''

def fmt_comp(row):
    v = row['total_comp']
    if row['porcentual'] == 1:
        return f"{v:,.1f}%"
    elif v >= 1000:
        return f"{int(v):,}"
    return f"{int(v)}"

def es_bueno(row):
    """
    Determina si el cambio es positivo para el programa.
    Captación (1) y Retención (2): subir = bueno
    Bajas (3): bajar = bueno (lógica invertida)
    """
    sube = row['porcentaje'] > 0
    if row['categoria'] == 3:
        return not sube   # En bajas, decrecer es bueno
    return sube

def color_num(row):
    if row['porcentaje'] == 0: return GRIS_NUM
    return VERDE_NUM if es_bueno(row) else ROJO

def color_barra(row):
    if row['porcentaje'] == 0: return GRIS
    return VERDE if es_bueno(row) else ROJO

def badge_html(row):
    sube  = row['porcentaje'] > 0
    pct   = abs(row['porcentaje'])
    flec  = '&#9650;' if sube else '&#9660;'
    if es_bueno(row):
        bg, tx = BG_VERDE, TX_VERDE
    else:
        bg, tx = BG_ROJO, TX_ROJO
    return (f'<span style="display:inline-block;background:{bg};color:{tx};'
            f'padding:3px 8px;border-radius:4px;font-size:10px;'
            f'font-family:monospace;">'
            f'{flec} {pct:.1f}%</span>')

print('✅ Funciones listas')
```

```python
# ── 8. GENERADORES DE BLOQUES HTML ────────────────────────────────────────────

F = "font-family:'Century Gothic',Arial,sans-serif;"
FM = "font-family:monospace;"

def bloque_grande(row):
    """Bloque KPI grande (Captación, 3 columnas)."""
    valor, sufijo = fmt_valor(row)
    cn    = color_num(row)
    badg  = badge_html(row)
    icono = ICONOS.get(row['id_indicador'], DEFAULT_ICONO)
    sup   = f'<sup style="font-size:13px;color:#AAA;font-weight:400;">{sufijo}</sup>' if sufijo else ''

    return f"""
    <td width="33%" style="background:#ffffff;padding:16px 14px;vertical-align:top;">
      <div style="margin-bottom:10px;">{icono}</div>
      <div style="font-size:28px;font-weight:800;line-height:1;margin-bottom:4px;
                  color:{cn};{F}">{valor}{sup}</div>
      <div style="font-size:9px;letter-spacing:1.5px;color:#AAA;text-transform:uppercase;
                  margin-bottom:10px;{F}">{row['nombre']}</div>
      {badg}
    </td>"""


def bloque_horizontal(row):
    """Bloque KPI horizontal (Retención y Bajas, 2 columnas)."""
    valor, sufijo = fmt_valor(row)
    comp  = fmt_comp(row)
    cn    = color_num(row)
    cb    = color_barra(row)
    sube  = row['porcentaje'] > 0
    flec  = '&#9650;' if sube else '&#9660;'
    pct   = abs(row['porcentaje'])
    cv    = VERDE if es_bueno(row) else ROJO
    sup   = f'<span style="font-size:12px;color:#CCC;">{sufijo}</span>' if sufijo else ''

    return f"""
    <td width="50%" style="background:#ffffff;padding:16px;vertical-align:top;">
      <table cellpadding="0" cellspacing="0" width="100%">
        <tr>
          <td width="3" style="background:{cb};border-radius:2px;">&nbsp;</td>
          <td style="padding-left:12px;">
            <div style="font-size:22px;font-weight:800;line-height:1;
                        margin-bottom:2px;color:{cn};{F}">{valor}{sup}</div>
            <div style="font-size:9px;letter-spacing:1px;color:#BBB;
                        text-transform:uppercase;margin-bottom:6px;{F}">{row['nombre']}</div>
            <div style="font-size:10px;color:{cv};{FM}">{flec} {pct:.1f}%</div>
            <div style="font-size:10px;color:#CCC;margin-top:2px;{F}">vs. {comp} en {ANIO_COMP}</div>
          </td>
        </tr>
      </table>
    </td>"""


def etiqueta_seccion(label):
    return f"""
    <tr>
      <td colspan="99" style="padding:16px 24px 6px;">
        <table cellpadding="0" cellspacing="0" width="100%">
          <tr>
            <td style="font-size:9px;letter-spacing:3px;color:{AZUL};opacity:0.6;
                       text-transform:uppercase;font-weight:500;
                       padding-right:12px;white-space:nowrap;{F}">{label}</td>
            <td style="border-top:1px solid {SEP};width:100%;"></td>
          </tr>
        </table>
      </td>
    </tr>"""


def grilla(filas_html, cols=3):
    """Envuelve las filas en tabla con separadores de 1px."""
    return f"""
    <tr>
      <td colspan="99" style="padding:0 24px 4px;">
        <table width="100%" cellpadding="0" cellspacing="1"
          style="background:{SEP};border-radius:10px;overflow:hidden;">
          {filas_html}
        </table>
      </td>
    </tr>"""

print('✅ Generadores listos')
```
